```python
import numpy as np
import matplotlib.pyplot as plt

# Task 1: Define Activation Functions

# Sigmoid Function
def sigmoid(x):
  return 1 / (1 + np.exp(-x))

# ReLU Function
def relu(x):
  return np.maximum(0, x)

# Tanh Function
def tanh(x):
  return np.tanh(x)
```

```python
# Task 2: Generate Outputs

# Input values
x_values = np.array([-5, -3, -1, 0, 1, 3, 5])

# Apply activation functions
sigmoid_output = sigmoid(x_values)
relu_output = relu(x_values)
tanh_output = tanh(x_values)

# Display outputs
print("Input Values:")
print(x_values)

print("\nSigmoid Output:")
print(sigmoid_output)

print("\nReLU Output:")
print(relu_output)

print("\nTanh Output:")
print(tanh_output)
```

```python
# Task 3: Visualize and Compare

# Generate continuous input values for plotting
x = np.linspace(-5, 5, 100)

# Calculate outputs
y_sigmoid = sigmoid(x)
y_relu = relu(x)
y_tanh = tanh(x)

# Plot all activation functions
plt.figure(figsize=(10, 6))

plt.plot(x, y_sigmoid, label='Sigmoid', color='blue')
plt.plot(x, y_relu, label='ReLU', color='green')
plt.plot(x, y_tanh, label='Tanh', color='red')

plt.title('Comparison of Activation Functions')
plt.xlabel('Input Value')
plt.ylabel('Output Value')
plt.legend()
plt.grid(True)
plt.show()
```
