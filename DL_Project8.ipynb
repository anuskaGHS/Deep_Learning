# Project 8: Gradient Descent Implementation

## Introduction

- During training, a neural network needs to **reduce its prediction error**
- This is done by updating the model parameters (weights and bias) using an algorithm called **Gradient Descent**
- Gradient Descent calculates how much the loss changes with respect to each parameter, and adjusts the parameters in the direction that reduces the loss
- In this project, Gradient Descent is implemented on the **Iris dataset** to observe how loss decreases over multiple iterations

## Task 1: Load Dataset and Initialize Parameters

- The **Iris dataset** is used with only **2 classes**, Setosa and Versicolor, to keep it a simple binary classification problem
- Input features are normalized using StandardScaler
- Weights are initialized to zeros and bias is set to zero

```python
import numpy as np
from sklearn.datasets import load_iris
from sklearn.preprocessing import StandardScaler
```

```python
# Load Iris dataset — use only first 2 classes (binary classification)
iris = load_iris()
X = iris.data[:100]       # Setosa and Versicolor only
y = iris.target[:100].reshape(-1, 1)   # Labels: 0 or 1
```

```python
# Normalize features
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)
```

```python
# Initialize parameters
weights = np.zeros((X_scaled.shape[1], 1))
bias = 0.0
```

```python
print("Dataset shape:", X_scaled.shape)
print("Labels shape:", y.shape)
print("Initial weights:", weights.T)
print("Initial bias:", bias)
```

## Task 2: Implement Gradient Descent

- **Sigmoid** activation is used to convert the output into a probability between 0 and 1
- **Binary Cross-Entropy** is used as the loss function
- In each iteration:
  - Forward pass - compute predictions
  - Calculate loss
  - Compute gradients (how much each parameter contributed to the error)
  - Update weights and bias using the gradients and learning rate

```python
# Sigmoid activation
def sigmoid(z):
    return 1 / (1 + np.exp(-z))

# Binary Cross-Entropy Loss
def compute_loss(y, y_pred):
    y_pred = np.clip(y_pred, 1e-10, 1 - 1e-10)
    return -np.mean(y * np.log(y_pred) + (1 - y) * np.log(1 - y_pred))
```

```python
# Gradient Descent
learning_rate = 0.1
iterations = 100
n = len(y)
loss_history = []

for i in range(iterations):

    # Forward pass
    z = np.dot(X_scaled, weights) + bias
    y_pred = sigmoid(z)

    # Calculate loss
    loss = compute_loss(y, y_pred)
    loss_history.append(loss)

    # Compute gradients
    error = y_pred - y
    dW = np.dot(X_scaled.T, error) / n
    db = np.sum(error) / n

    # Update parameters
    weights = weights - learning_rate * dW
    bias = bias - learning_rate * db

print("Training complete.")
print(f"Initial Loss: {round(loss_history[0], 4)}")
print(f"Final Loss: {round(loss_history[-1], 4)}")
print(f"Final Weights: {np.round(weights.T, 4)}")
print(f"Final Bias: {round(bias, 4)}")
```

## Task 3: Analyze How Loss Decreases Over Iterations

- The loss at every iteration is stored in `loss_history`
- We print the loss at key checkpoints to observe the reduction

```python
print(f"{'Iteration':<15} {'Loss'}")
print("-" * 30)
checkpoints = [0, 9, 19, 39, 59, 79, 99]
for i in checkpoints:
    print(f"{i+1:<15} {round(loss_history[i], 4)}")
```

## Conclusion

- Gradient Descent successfully reduced the loss over 100 iterations on the Iris dataset
- The loss started high because the weights were randomly initialized to zero and the model had not learned anything yet
- With each iteration, the gradients guided the weights and bias in the direction that reduced the error
- By the final iteration, the loss dropped significantly showing that the model parameters were being updated correctly
- This confirms that Gradient Descent is the core mechanism behind neural network training
